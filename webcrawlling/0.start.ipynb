html 속에 있는 요소를 꺼내는 방법
1. 요소를 직접 지정 > find("요소명", id="id명")
2. CSS의 요소 검색 방법 > select("선택자")

선택자란, css에서 요소를 지정하는 방법.

CSS에서 #은 id, .은 class를 나타냅니다.  "#subject" ".contents"
자식은 > 로 지정합니다.  "span > b" (span은 부모, b는 자식)
여러 개 요소 목록 ,로 나열합니다.  "div, h1"

- select()는 지정한 선택자의 요소를 모두 가져옵니다. [리스트로]
- select_one()은 지정한 선택자 요소 중 첫번째 요소만 반환합니다.

```python
soup.select_one('div#subject') #id로 꺼내기
```

```python
soup.select_one('div.contents') #class로 꺼내기
```

```python
b_tag = soup.select_one("div > b")  #contents 클래스 안에 있는 b 태그
print(b_tag.text)
```

```python
soup.select_one("div > b").text
```

```python
url = 'https://finance.naver.com/marketindex/'
market_index = requests.get(url)
market_index.text[:100]  #HTML의 처음 100글자 출력
```

```python
# USD/KRW 환율 값 찾기

import requests
from bs4 import BeautifulSoup

url = "https://finance.naver.com/marketindex/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# 'USD' 텍스트가 있는 환율 박스 선택
usd_krw_element = soup.select_one("a.head.usd > div.head_info > span.value")

usd_krw = usd_krw_element.get_text().strip()

print('USD/KRW 환율:', usd_krw)
```

```python
# USD/JPY 환율 값 찾기

import requests
from bs4 import BeautifulSoup

url = "https://finance.naver.com/marketindex/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# 'JPY' 텍스트가 있는 환율 박스 선택
usd_jpy_element = soup.select_one("a.head.jpy_usd > div > span.value")

usd_jpy = usd_jpy_element.get_text().strip()

print('USD/JPY 환율:', usd_jpy)
```
