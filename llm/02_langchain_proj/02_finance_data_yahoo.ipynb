```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
load_dotenv()

llm = ChatOpenAI(model='gpt-4o', temperature=0)
```

llm 보고서 생성하기

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
  ("system","""
  Want assistance provided by qualified individuals enabled with experience on understanding charts using technical analysis tools while interpreting macroeconomic environment prevailing across world consequently assisting customers acquire long term advantages requires clear verdicts therefore seeking same through informed predictions written down precisely! First statement contains following content- “Can you tell us what future stock market looks like based upon current conditions ?"."""),
  ("user","""
      {company} 주식에 투자를 해도 될까요? 마크다운 형식의 투자 보고서를 한글로 작성해주세요.""")
])

output_parser = StrOutputParser()

chain = prompt | llm | output_parser

company = "Microsoft"

response = chain.invoke({"company":company})
print(response)
```

# 기본 정보 출력

```python
import yfinance as yf

# Microsoft 주식 심볼 (MSFT)
ticker = 'MSFT'

# 티커 정보 가져오기
msft = yf.Ticker(ticker)

# 기본 정보 조회
basic_info = {
    "회사명": msft.info.get('longName'),
    "산업": msft.info.get('industry'),
    "부문": msft.info.get('sector'),
    "시가총액": msft.info.get('marketCap'),
    "총 발행 주식 수": msft.info.get('sharesOutstanding'),
    "본사 위치": msft.info.get('country'),
    "CEO": msft.info.get('ceo'),  # 일부 주식에서 사용 가능
    "웹사이트": msft.info.get('website'),
}

# 기본 정보 출력
for key, value in basic_info.items():
    print(f"{key}: {value}")
```

# 14. 자사주 매입 (현금흐름에서 확인)

```python
# ────────────────────────────────────────────
# 14. 자사주 매입 (현금흐름에서 확인)
# ────────────────────────────────────────────
print("\n" + "=" * 60)
print("14. 자사주 매입 현황")
print("=" * 60)
cf = msft.cashflow
if 'Repurchase Of Capital Stock' in cf.index:
    print(cf.loc['Repurchase Of Capital Stock'])
elif 'Common Stock Repurchase' in cf.index:
    print(cf.loc['Common Stock Repurchase'])
else:
    buyback_rows = [r for r in cf.index if 'Repurchase' in str(r) or 'Buyback' in str(r)]
    print(cf.loc[buyback_rows] if buyback_rows else "데이터 없음")
```
