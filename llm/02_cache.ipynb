# RedisSemanticCache   

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
REDIS_URL = os.environ['REDIS_URL']
```

```python
from langchain_core.globals import set_llm_cache
from langchain_redis import RedisSemanticCache          # 변경
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_core.messages import HumanMessage
```

```python
semantic_cache = RedisSemanticCache(
    redis_url=REDIS_URL,
    embeddings=OpenAIEmbeddings(model="text-embedding-3-small"),  # 파라미터명 변경
    distance_threshold=0.15,
    ttl=3600                                        # 파라미터명 변경
)
set_llm_cache(semantic_cache)
```

```python
llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
response = llm.invoke([HumanMessage(content="코스피 지수란 무엇인가요? 한 문장으로 답해주세요.")])
print(response.content)
```

```python
response = llm.invoke([HumanMessage(content="코스피가 뭔지 간단히 설명해줘.")])
print(response.content)
```

```python
messages = [
  ("system", "당신은 주식 투자 전문가이다."),
  ("human", "현재 삼성전자 주가를 분석해줘")
]

response = llm.invoke(messages)
print(response.content)
```

```python
import numpy as np

emb = OpenAIEmbeddings(model="text-embedding-3-small")

q1 = "코스피 지수란 무엇인가요? 한 문장으로 답해주세요."
q2 = "코스피가 뭔지 간단히 설명해줘."
q3 = "코스닥 지수란 무엇인가요?"

v1, v2, v3 = emb.embed_documents([q1, q2, q3])

def cosine_dist(a, b):
    return 1 - np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(f"코스피↔코스피 유사질문 거리: {cosine_dist(v1, v2):.4f}")  # 작아야 함 (캐시 히트)
print(f"코스피↔코스닥 거리:          {cosine_dist(v1, v3):.4f}")  # 커야 함 (캐시 미스)
```

```python
semantic_cache = RedisSemanticCache(
    redis_url=REDIS_URL,
    embeddings=OpenAIEmbeddings(model="text-embedding-3-small"),
    distance_threshold=0.4,  # 0.3612보다 커야 코스피 유사질문 캐시 히트
    ttl=3600
)
set_llm_cache(semantic_cache)

llm = ChatOpenAI(model="gpt-4o-mini")

questions = [
    "코스피 지수란 무엇인가요? 한 문장으로 답해주세요.",  # 최초 질문 (캐시 미스)
    "코스피가 뭔지 간단히 설명해줘.",                    # 유사 질문 (캐시 히트 기대)
    "코스닥 지수란 무엇인가요?",                         # 다른 질문 (캐시 미스 기대, but 안됨)
]

for q in questions:
    start = time.time()
    answer = llm.invoke(q)
    elapsed = time.time() - start
    
    print(f"질문: {q}")
    print(f"답변: {answer.content}")
    print(f"소요시간: {elapsed:.2f}초  {'← 캐시 히트 (빠름)' if elapsed < 0.5 else '← API 호출 (느림)'}")
    print("-" * 60)
```
