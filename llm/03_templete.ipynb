# ChatPromptTemplate

load, save 함수는 deprecated, 직접 json 파일을 저장하고 로드합니다.

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 베테랑 여행 가이드로, 고객 최적의 맞춤 여행 일정 수립을 도와줍니다."),
    ("human", "{when}에 {place}로 {travel}을 계획하고 있습니다."),
    ("ai",    "제가 바로 {place} 전문 여행 가이드입니다. 어떤 도움이 필요하신가요?"),
    ("human", "{question}")
])
```

# 프롬프트 구성 > LCEL 파이프라인 구성 > invoke

금융도메인 프롬프트 : system(역할, 페르소나), user, ai(퓨샷러닝), user로 구성합니다.

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 포트폴리오 투자 전문가 레이 달리오입니다. 어디에 투자할지 조언을 구하는 지인에게 현 시점에 맞는 포트폴리오를 조언해줍니다.."),
    ("user", "현금 {money}로 {포트폴리오}별 {투자비중}을 고려하고 있습니다."),
    ("ai",    "제가 바로 {place} 전문 여행 가이드입니다. 어떤 도움이 필요하신가요?"),
    ("human", "{question}")
])



from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 베테랑 여행 가이드로, 고객 최적의 맞춤 여행 일정 수립을 도와줍니다."),
    ("human", "{when}에 {place}로 {travel}을 계획하고 있습니다."),
    ("ai",    "제가 바로 {place} 전문 여행 가이드입니다. 어떤 도움이 필요하신가요?"),
    ("human", "{question}")
])
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# ================================================================
# 1. 금융 도메인 프롬프트 구성
# ================================================================

chat_template = ChatPromptTemplate.from_messages([
    ("system",  "당신은 레이달리오로, 고객에게 {포트폴리오} 전략을 쉽고 명확하게 설명해 드립니다."),
    ("human",   "저는 {level} 투자자로, {포트폴리오}에 관심이 많습니다."),
    ("ai",      "안녕하세요! {포트폴리오} 전문가 레이달리오입니다. {level} 눈높이에 맞춰 설명해 드리겠습니다. 무엇이 궁금하신가요?"),
    ("human",   "{question}")
])

# ================================================================
# 2. LCEL 파이프라인 구성
# ================================================================

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.3)
output_parser = StrOutputParser()

chain = chat_template | llm | output_parser

# ================================================================
# 3. invoke
# ================================================================

response = chain.invoke({
    "포트폴리오"  : "여러 금융상품(채권, 주식, 원자재 선물 등)",
    "level"   : "중급자",
    "question": "현금 1억으로 연평균 6% 수익을 낼 수 있는 포트폴리오를 추천해주세요."
})

print(response)
```
