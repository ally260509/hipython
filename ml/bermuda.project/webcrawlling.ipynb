# 2단계: 자동 이동 실행

```python
# 클릭 보조함수 만들기

wait = WebDriverWait(driver, 15)

def safe_click(candidates, name):
    last_error = None
    
    for by, selector in candidates:
        try:
            elem = wait.until(EC.presence_of_element_located((by, selector)))
            driver.execute_script("arguments[0].scrollIntoView({block:'center'});", elem)
            time.sleep(1)
            
            try:
                wait.until(EC.element_to_be_clickable((by, selector)))
                elem.click()
            except:
                driver.execute_script("arguments[0].click();", elem)
            
            print(f"{name} 클릭 완료")
            time.sleep(2)
            return
        
        except Exception as e:
            last_error = e
    
    raise last_error
```

```python
# 엔카 메인 이동
driver.get("https://www.encar.com/index.do")
time.sleep(3)

print("현재 URL:", driver.current_url)
print("페이지 제목:", driver.title)
```

```python
# 국산 -> 현대 -> 그렌저 클릭 

# 1) 국산 클릭 후보
domestic_candidates = [
    (By.XPATH, "//a[normalize-space()='국산']"),
    (By.XPATH, "//button[normalize-space()='국산']"),
    (By.XPATH, "//*[normalize-space()='국산']/ancestor::*[self::a or self::button][1]")
]

# 2) 현대 클릭 후보
hyundai_candidates = [
    (By.XPATH, "//a[normalize-space()='현대']"),
    (By.XPATH, "//button[normalize-space()='현대']"),
    (By.XPATH, "//label[normalize-space()='현대']"),
    (By.XPATH, "//*[normalize-space()='현대']/ancestor::*[self::a or self::button or self::label][1]")
]

# 3) 그랜저 클릭 후보
grandeur_candidates = [
    (By.XPATH, "//a[normalize-space()='그랜저']"),
    (By.XPATH, "//button[normalize-space()='그랜저']"),
    (By.XPATH, "//label[normalize-space()='그랜저']"),
    (By.XPATH, "//*[normalize-space()='그랜저']/ancestor::*[self::a or self::button or self::label][1]")
]

# 클릭 실행
safe_click(domestic_candidates, "국산")
safe_click(hyundai_candidates, "현대")
safe_click(grandeur_candidates, "그랜저")

print("2단계 이동 완료")
print("현재 URL:", driver.current_url)
print("페이지 제목:", driver.title)
```

# 4단계: 카드 텍스트 전체를 먼저 수집

```python
# 수집된 카드 수 및 첫 번째 카드 예시

cards = []

for link in detail_links:
    try:
        card = link.find_element(By.XPATH, "./..")   # 상위 1단계 = 카드 1개
        title = link.get_attribute("title")
        href = link.get_attribute("href")
        text = card.text.strip()
        
        cards.append({
            "title": title,
            "detail_url": href,
            "card_text": text
        })
    except Exception as e:
        print("에러 발생:", e)

print("수집된 카드 수:", len(cards))
print("\n첫 번째 카드 예시:")
print(cards[0])
```

```python
# card_text 컬럼으로 분리하기

import re
import pandas as pd

def clean_line(line):
    line = line.strip()
    line = re.sub(r'^[·•]\s*', '', line)   # 앞의 점 제거
    return line

def parse_card_text(text):
    lines = [clean_line(x) for x in text.split("\n")]
    lines = [x for x in lines if x and x != "찜"]   # 빈 줄, '찜' 제거
    
    result = {
        "car_name": "",
        "year_info": "",
        "mileage": "",
        "fuel": "",
        "region": "",
        "price_info": ""
    }
    
    if len(lines) >= 1:
        result["car_name"] = lines[0]
    if len(lines) >= 2:
        result["year_info"] = lines[1]
    if len(lines) >= 3:
        result["mileage"] = lines[2]
    if len(lines) >= 4:
        result["fuel"] = lines[3]
    if len(lines) >= 5:
        result["region"] = lines[4]
    if len(lines) >= 6:
        result["price_info"] = " ".join(lines[5:])   # 뒤에 여러 줄이 있어도 합치기
    
    return result


parsed_cards = []

for card in cards:
    parsed = parse_card_text(card["card_text"])
    parsed["detail_url"] = card["detail_url"]
    parsed_cards.append(parsed)

list_df = pd.DataFrame(parsed_cards)

print("데이터 개수:", len(list_df))
list_df.head(10)
```

```python
# 상세 본문 후보 찾기

candidates = []

elements = driver.find_elements(By.XPATH, "//*[self::div or self::section or self::article]")

for elem in elements:
    try:
        if not elem.is_displayed():
            continue
        
        text = elem.text.strip()
        
        if len(text) < 100:
            continue
        
        if "기본정보" in text and ("옵션정보" in text or "판매자 정보" in text):
            candidates.append({
                "tag": elem.tag_name,
                "id": elem.get_attribute("id"),
                "class": elem.get_attribute("class"),
                "text": text
            })
    except:
        pass

print("후보 개수:", len(candidates))

for i, c in enumerate(candidates[:5], 1):
    print(f"\n===== 후보 {i} =====")
    print("tag   :", c["tag"])
    print("id    :", c["id"])
    print("class :", c["class"])
    print(c["text"][:1500])
```

```python
# 현재 페이지에 차량명이 보이는지 직접 검색

from selenium.webdriver.common.by import By

# 현재 문맥에서 '그랜저'가 들어간 보이는 요소 찾기
elems = driver.find_elements(By.XPATH, "//*[contains(normalize-space(.), '그랜저')]")

visible_elems = []
for e in elems:
    try:
        if e.is_displayed():
            txt = e.text.strip()
            if txt:
                visible_elems.append(txt)
    except:
        pass

# 중복 제거
visible_unique = []
seen = set()
for t in visible_elems:
    if t not in seen:
        visible_unique.append(t)
        seen.add(t)

print("현재 문맥에서 보이는 '그랜저' 요소 수:", len(visible_unique))
for i, t in enumerate(visible_unique[:10], 1):
    print(f"\n[{i}]")
    print(t[:300])
```

```python
# iframe 안에서 차량명 찾기

driver.switch_to.default_content()

found = False

iframes = driver.find_elements(By.TAG_NAME, "iframe")

for idx in range(len(iframes)):
    driver.switch_to.default_content()
    iframes = driver.find_elements(By.TAG_NAME, "iframe")
    
    try:
        driver.switch_to.frame(iframes[idx])
        elems = driver.find_elements(By.XPATH, "//*[contains(normalize-space(.), '그랜저')]")
        
        visible_texts = []
        for e in elems:
            try:
                if e.is_displayed():
                    txt = e.text.strip()
                    if txt:
                        visible_texts.append(txt)
            except:
                pass
        
        unique_texts = []
        seen = set()
        for t in visible_texts:
            if t not in seen:
                unique_texts.append(t)
                seen.add(t)
        
        if unique_texts:
            print(f"[성공] iframe {idx} 안에서 '그랜저' 발견")
            for i, t in enumerate(unique_texts[:10], 1):
                print(f"\n[{i}]")
                print(t[:300])
            found = True
            break
    
    except Exception as e:
        print(f"iframe {idx} 확인 중 에러:", e)

if not found:
    print("iframe 안에서도 '그랜저'를 찾지 못했어.")
```

```python
# 차량명/가격/km 같은 핵심 요소만 좁게 찾기

keywords = ["그랜저", "만원", "km", "주행거리", "연료", "변속기", "배기량"]

elems = driver.find_elements(
    By.XPATH,
    "//*[self::h1 or self::h2 or self::h3 or self::div or self::span or self::li or self::dd or self::dt or self::strong]"
)

matches = []

for e in elems:
    try:
        if not e.is_displayed():
            continue
        
        txt = e.text.strip()
        if not txt:
            continue
        
        # 너무 긴 블록 제외
        if len(txt) > 300:
            continue
        
        # 키워드 포함 여부
        if any(k in txt for k in keywords):
            matches.append(txt)
    except:
        pass

# 중복 제거
unique_matches = []
seen = set()
for t in matches:
    if t not in seen:
        unique_matches.append(t)
        seen.add(t)

print("핵심 후보 수:", len(unique_matches))
for i, t in enumerate(unique_matches[:30], 1):
    print(f"\n[{i}]")
    print(t)
```

```python
# JS로 실제 보이는 텍스트 노드 추출

visible_texts = driver.execute_script("""
const results = [];
const walker = document.createTreeWalker(document.body, NodeFilter.SHOW_TEXT);

while (walker.nextNode()) {
    const node = walker.currentNode;
    const text = node.textContent.trim();
    if (!text) continue;

    const parent = node.parentElement;
    if (!parent) continue;

    const style = window.getComputedStyle(parent);
    const rect = parent.getBoundingClientRect();

    if (style.display === 'none' || style.visibility === 'hidden') continue;
    if (rect.width === 0 || rect.height === 0) continue;

    results.push(text);
}

return [...new Set(results)].slice(0, 300);
""")

print("보이는 텍스트 수:", len(visible_texts))
for i, t in enumerate(visible_texts[:50], 1):
    print(f"\n[{i}] {t}")
```
