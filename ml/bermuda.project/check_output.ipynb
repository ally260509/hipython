```python
from pathlib import Path
import pandas as pd

BATCH_CSV = Path("encar_kia_all/output/kia_detail_batch.csv")
SUCCESS_LOG_JSONL = Path("encar_kia_all/output/kia_detail_success_log.jsonl")
ERROR_LOG_JSONL = Path("encar_kia_all/output/kia_detail_error_log.jsonl")

print("CSV 경로:", BATCH_CSV.resolve())
print("SUCCESS LOG 경로:", SUCCESS_LOG_JSONL.resolve())
print("ERROR LOG 경로:", ERROR_LOG_JSONL.resolve())

print("CSV 존재:", BATCH_CSV.exists())
print("SUCCESS LOG 존재:", SUCCESS_LOG_JSONL.exists())
print("ERROR LOG 존재:", ERROR_LOG_JSONL.exists())

if BATCH_CSV.exists():
    try:
        df = pd.read_csv(BATCH_CSV)
        print("CSV 행 수:", len(df))
        if len(df) > 0:
            print(df[["매물ID", "차량명", "현재가격_만원"]].tail(5))
    except Exception as e:
        print("CSV 읽기 중 오류:", e)

if SUCCESS_LOG_JSONL.exists():
    try:
        with open(SUCCESS_LOG_JSONL, "r", encoding="utf-8") as f:
            success_lines = f.readlines()
        print("SUCCESS LOG 줄 수:", len(success_lines))
    except Exception as e:
        print("SUCCESS LOG 읽기 중 오류:", e)

if ERROR_LOG_JSONL.exists():
    try:
        with open(ERROR_LOG_JSONL, "r", encoding="utf-8") as f:
            error_lines = f.readlines()
        print("ERROR LOG 줄 수:", len(error_lines))
    except Exception as e:
        print("ERROR LOG 읽기 중 오류:", e)
```
