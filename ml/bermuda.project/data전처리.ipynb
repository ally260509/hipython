```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Admin\hipython\ml\bermuda.project\data\merged_all_brands_by_my_template (1).csv")
```

```python
def normalize_model(model_name):
    if pd.isna(model_name):
        return model_name
    
    model_name = str(model_name)

    # 대표 규칙 정의
    if "SM7" in model_name:
        return "SM7"
    elif "SM5" in model_name:
        return "SM5"
    elif "SM6" in model_name:
        return "SM6"
    elif "그랜저" in model_name:
        return "그랜저"
    elif "쏘나타" in model_name:
        return "쏘나타"
    
    # 필요시 계속 확장 가능
    return model_name  # 매칭 안되면 원본 유지
```

```python
df["모델_그룹"] = df["모델"].apply(normalize_model)
```

```python
model_freq = (
    df["모델_그룹"]
    .value_counts()
    .reset_index()
)

model_freq.columns = ["모델_그룹", '빈도']
print(model_freq)
```

```python
import pandas as pd
import sys

CSV_PATH = r'C:\Users\Admin\hipython\ml\bermuda.project\data\merged_all_brands_by_my_template (1).csv'

# ──────────────────────────────────────────────
# 모델 → 그룹명 매핑 (직접 지정 방식)
# 매핑에 없는 모델은 원래 이름 그대로 유지됨
# ──────────────────────────────────────────────
MODEL_GROUP_MAP = {

    # ═══ 현대 (Hyundai) ═══

    # 그랜저
    '그랜저': '그랜저',
    '그랜저 HG': '그랜저',
    '그랜저 IG': '그랜저',
    '그랜저 (GN7)': '그랜저',
    '더 뉴 그랜저 IG': '그랜저',
    '그랜저 HG 하이브리드': '그랜저',
    '그랜저 IG 하이브리드': '그랜저',
    '그랜저 하이브리드 (GN7)': '그랜저',
    '그랜저 뉴 럭셔리': '그랜저',
    '더 럭셔리 그랜저': '그랜저',
    '더 뉴 그랜저 IG 하이브리드': '그랜저',
    '그랜저 TG': '그랜저',

    # 쏘나타
    '쏘나타': '쏘나타',
    '쏘나타 (DN8)': '쏘나타',
    'LF 쏘나타': '쏘나타',
    'YF 쏘나타': '쏘나타',
    'NF 쏘나타': '쏘나타',
    'NF 쏘나타 트랜스폼': '쏘나타',
    '뉴 EF 쏘나타': '쏘나타',
    '쏘나타 디 엣지(DN8)': '쏘나타',
    '쏘나타 뉴 라이즈': '쏘나타',
    '쏘나타 더 브릴리언트': '쏘나타',
    '쏘나타 하이브리드': '쏘나타',
    '쏘나타 하이브리드 (DN8)': '쏘나타',
    '쏘나타 디 엣지 하이브리드(DN8)': '쏘나타',
    '쏘나타 뉴 라이즈 하이브리드': '쏘나타',
    'LF 쏘나타 하이브리드': '쏘나타',

    # 아반떼
    '아반떼': '아반떼',
    '아반떼 MD': '아반떼',
    '아반떼 AD': '아반떼',
    '아반떼 HD': '아반떼',
    '아반떼 XD': '아반떼',
    '아반떼 (CN7)': '아반떼',
    '더 뉴 아반떼': '아반떼',
    '더 뉴 아반떼 AD': '아반떼',
    '더 뉴 아반떼 (CN7)': '아반떼',
    '아반떼 하이브리드': '아반떼',
    '아반떼 하이브리드 (CN7)': '아반떼',
    '더 뉴 아반떼 하이브리드 (CN7)': '아반떼',

    # 투싼
    '투싼': '투싼',
    '투싼 ix': '투싼',
    '뉴 투싼 ix': '투싼',
    '올 뉴 투싼': '투싼',
    '투싼 (NX4)': '투싼',
    '더 뉴 투싼 (NX4)': '투싼',
    '투싼 하이브리드 (NX4)': '투싼',
    '더 뉴 투싼 하이브리드 (NX4)': '투싼',

    # 싼타페
    '싼타페 (MX5)': '싼타페',
    '싼타페 TM': '싼타페',
    '싼타페 DM': '싼타페',
    '싼타페 CM': '싼타페',
    '더 뉴 싼타페': '싼타페',
    '싼타페 더 프라임': '싼타페',

    # 팰리세이드
    '팰리세이드': '팰리세이드',
    '더 뉴 팰리세이드': '팰리세이드',
    '팰리세이드 (LX3)': '팰리세이드',

    # 코나
    '코나': '코나',
    '더 뉴 코나': '코나',
    '코나 (SX2)': '코나',
    '코나 하이브리드': '코나',
    '코나 하이브리드 (SX2)': '코나',
    '코나 일렉트릭': '코나',
    '코나 일렉트릭 (SX2)': '코나',
    '더 뉴 코나 하이브리드': '코나',

    # 스타렉스 / 그랜드 스타렉스
    '그랜드 스타렉스': '스타렉스',
    '더 뉴 그랜드 스타렉스': '스타렉스',
    '스타렉스': '스타렉스',
    '스타렉스 점보': '스타렉스',

    # 스타리아
    '스타리아': '스타리아',

    # 아이오닉 (1세대 하이브리드/전기)
    '아이오닉 하이브리드': '아이오닉',
    '더 뉴 아이오닉 하이브리드': '아이오닉',
    '아이오닉 일렉트릭': '아이오닉',
    '더 뉴 아이오닉 일렉트릭': '아이오닉',

    # 아이오닉5
    '아이오닉5': '아이오닉5',
    '더 뉴 아이오닉5': '아이오닉5',

    # 아이오닉6
    '아이오닉6': '아이오닉6',

    # 아이오닉9
    '아이오닉9': '아이오닉9',

    # 벨로스터
    '벨로스터': '벨로스터',
    '벨로스터 (JS)': '벨로스터',
    '더 뉴 벨로스터': '벨로스터',

    # i30
    'i30': 'i30',
    'i30 (PD)': 'i30',
    'i30(신형)': 'i30',
    '더 뉴 i30': 'i30',

    # i40
    'i40': 'i40',
    'i40 살룬': 'i40',
    '더 뉴 i40': 'i40',
    '더 뉴 i40 살룬': 'i40',

    # 넥쏘
    '넥쏘': '넥쏘',

    # 베뉴
    '베뉴': '베뉴',

    # 맥스크루즈
    '맥스크루즈': '맥스크루즈',
    '더 뉴 맥스크루즈': '맥스크루즈',

    # 에쿠스
    '에쿠스': '에쿠스',
    '에쿠스(신형)': '에쿠스',

    # 베라크루즈
    '베라크루즈': '베라크루즈',

    # 아슬란
    '아슬란': '아슬란',

    # 엑센트
    '엑센트(신형)': '엑센트',

    # ═══ 기아 (Kia) ═══

    # K5
    'K5': 'K5',
    'K5 2세대': 'K5',
    'K5 3세대': 'K5',
    '더 뉴 K5': 'K5',
    '더 뉴 K5 2세대': 'K5',
    '더 뉴 K5 3세대': 'K5',
    'K5 하이브리드': 'K5',
    'K5 하이브리드 2세대': 'K5',
    'K5 하이브리드 3세대': 'K5',
    '더 뉴 K5 하이브리드 3세대': 'K5',

    # K7
    'K7': 'K7',
    '더 뉴 K7': 'K7',
    '올 뉴 K7': 'K7',
    'K7 프리미어': 'K7',
    'K7 하이브리드': 'K7',
    'K7 프리미어 하이브리드': 'K7',
    '더 프레스티지 K7': 'K7',
    '올 뉴 K7 하이브리드': 'K7',

    # K3
    'K3': 'K3',
    '올 뉴 K3': 'K3',
    '더 뉴 K3': 'K3',
    '더 뉴 K3 2세대': 'K3',
    'K3 쿱': 'K3',

    # K8
    'K8': 'K8',
    '더 뉴 K8': 'K8',
    'K8 하이브리드': 'K8',
    '더 뉴 K8 하이브리드': 'K8',

    # K9
    'K9': 'K9',
    '더 K9': 'K9',
    '더 뉴 K9': 'K9',
    '더 뉴 K9 2세대': 'K9',

    # 카니발
    '카니발': '카니발',
    '올 뉴 카니발': '카니발',
    '카니발 R': '카니발',
    '더 뉴 카니발': '카니발',
    '카니발 4세대': '카니발',
    '더 뉴 카니발 4세대': '카니발',
    '그랜드 카니발': '카니발',
    '뉴카니발': '카니발',

    # 쏘렌토
    '쏘렌토 4세대': '쏘렌토',
    '올 뉴 쏘렌토': '쏘렌토',
    '더 뉴 쏘렌토': '쏘렌토',
    '더 뉴 쏘렌토 4세대': '쏘렌토',
    '쏘렌토 R': '쏘렌토',
    '뉴 쏘렌토 R': '쏘렌토',

    # 스포티지
    '스포티지 4세대': '스포티지',
    '스포티지 5세대': '스포티지',
    '스포티지 R': '스포티지',
    '더 뉴 스포티지 R': '스포티지',
    '스포티지 더 볼드': '스포티지',
    '더 뉴 스포티지 5세대': '스포티지',
    '스포티지 5세대 하이브리드': '스포티지',
    '더 뉴 스포티지 5세대 하이브리드': '스포티지',
    'New 스포티지': '스포티지',

    # 셀토스
    '셀토스': '셀토스',
    '더 뉴 셀토스': '셀토스',

    # 모하비
    '모하비': '모하비',
    '더 뉴 모하비': '모하비',
    '모하비 더 마스터': '모하비',

    # 캐스퍼
    '캐스퍼': '캐스퍼',
    '더 뉴 캐스퍼': '캐스퍼',
    '캐스퍼 일렉트릭': '캐스퍼',

    # 니로
    '니로': '니로',
    '더 뉴 니로': '니로',
    '디 올 뉴 니로': '니로',
    '니로 EV': '니로',
    '디 올 뉴 니로 EV': '니로',
    '니로 플러스': '니로',

    # 레이
    '레이': '레이',
    '더 뉴 레이': '레이',
    '더 뉴 기아 레이': '레이',
    '더 뉴 기아 레이 EV': '레이',

    # 모닝
    '모닝': '모닝',
    '올 뉴 모닝': '모닝',
    '올 뉴 모닝 (JA)': '모닝',
    '모닝 어반 (JA)': '모닝',
    '더 뉴 모닝': '모닝',
    '더 뉴 모닝 (JA)': '모닝',
    '뉴모닝': '모닝',

    # 스팅어
    '스팅어': '스팅어',
    '스팅어 마이스터': '스팅어',

    # EV 시리즈 (기아)
    'EV3': 'EV3',
    'EV4': 'EV4',
    'EV5': 'EV5',
    'EV6': 'EV6',
    '더 뉴 EV6': 'EV6',
    'EV9': 'EV9',

    # 포르테
    '포르테': '포르테',
    '포르테 쿱': '포르테',

    # 쏘울
    '쏘울': '쏘울',
    '올 뉴 쏘울': '쏘울',
    '쏘울 부스터': '쏘울',
    '쏘울 부스터 EV': '쏘울',
    '쏘울 EV': '쏘울',

    # 카렌스
    '뉴카렌스': '카렌스',
    '올 뉴 카렌스': '카렌스',
    '더 뉴 카렌스': '카렌스',

    # 기타
    '스토닉': '스토닉',
    '쏠라티': '쏠라티',
    '타스만': '타스만',

    # ═══ 제네시스 (Genesis) ═══

    # G80
    'G80': 'G80',
    'G80 (RG3)': 'G80',
    '일렉트리파이드 G80 (RG3)': 'G80',

    # G90
    'G90': 'G90',
    'G90 (RS4)': 'G90',
    'EQ900': 'G90',

    # GV70
    'GV70': 'GV70',
    '일렉트리파이드 GV70': 'GV70',

    # GV80
    'GV80': 'GV80',
    'GV80 쿠페': 'GV80',

    # GV60
    'GV60': 'GV60',

    # G70
    'G70': 'G70',
    '더 뉴 G70': 'G70',
    '더 뉴 G70 슈팅브레이크': 'G70',

    # 제네시스 쿠페 / 구형 제네시스 세단
    '제네시스': '제네시스 (구형)',
    '제네시스 DH': '제네시스 (구형)',
    '제네시스 쿠페': '제네시스 쿠페',
    '더 뉴 제네시스 쿠페': '제네시스 쿠페',

    # ═══ KG모빌리티 (SsangYong) ═══

    # 렉스턴
    'G4 렉스턴': '렉스턴',
    '올 뉴 렉스턴': '렉스턴',
    '렉스턴 W': '렉스턴',
    '렉스턴 뉴 아레나': '렉스턴',
    '슈퍼 렉스턴': '렉스턴',

    # 렉스턴 스포츠 / 스포츠 칸
    '렉스턴 스포츠': '렉스턴 스포츠',
    '더 뉴 렉스턴 스포츠': '렉스턴 스포츠',
    '렉스턴 스포츠 칸': '렉스턴 스포츠',
    '더 뉴 렉스턴 스포츠 칸': '렉스턴 스포츠',

    # 티볼리
    '티볼리': '티볼리',
    '티볼리 에어': '티볼리',
    '티볼리 아머': '티볼리',
    '베리 뉴 티볼리': '티볼리',
    '더 뉴 티볼리': '티볼리',
    '더 뉴 티볼리 에어': '티볼리',

    # 코란도
    '뷰티풀 코란도': '코란도',
    '뉴코란도 C': '코란도',
    '코란도 C': '코란도',
    '뉴 스타일 코란도 C': '코란도',
    '뉴코란도': '코란도',
    '코란도 투리스모': '코란도 투리스모',  # 대형 버스형 → 별도 유지
    '코란도 스포츠': '코란도 스포츠',    # 픽업트럭 → 별도 유지
    '더 뉴 코란도 스포츠': '코란도 스포츠',
    '코란도 지프': '코란도 지프',

    # 토레스
    '토레스': '토레스',
    '더 뉴 토레스': '토레스',
    '토레스 EVX': '토레스',

    # 체어맨
    '뉴체어맨 W': '체어맨',
    '뉴체어맨': '체어맨',
    '체어맨 H 뉴 클래식': '체어맨',

    # 무쏘
    '뉴무쏘': '무쏘',

    # 액티언
    '액티언': '액티언',
    '액티언 2세대': '액티언',
    '액티언 스포츠': '액티언 스포츠',

    # ═══ 르노코리아 (Renault Korea) ═══

    # SM7 ← 사용자 예시 그대로
    'SM7': 'SM7',
    'SM7 노바': 'SM7',
    'SM7 New Art': 'SM7',
    'All New SM7': 'SM7',

    # SM6
    'SM6': 'SM6',
    '더 뉴 SM6': 'SM6',

    # SM5
    'SM5': 'SM5',
    'SM5 노바': 'SM5',
    'SM5 뉴 임프레션': 'SM5',
    '뉴SM5 플래티넘': 'SM5',
    '뉴SM5(신형)': 'SM5',

    # SM3
    'SM3': 'SM3',
    'SM3 네오': 'SM3',
    '뉴SM3': 'SM3',
    'SM3 Z.E.': 'SM3',

    # QM6
    'QM6': 'QM6',
    '더 뉴 QM6': 'QM6',

    # QM5
    'QM5': 'QM5',
    'QM5 네오': 'QM5',
    '뉴QM5': 'QM5',

    # QM3
    'QM3': 'QM3',
    '뉴QM3': 'QM3',

    # XM3 / 그랑 콜레오스 / 기타
    'XM3': 'XM3',
    '그랑 콜레오스': '그랑 콜레오스',
    '마스터': '마스터',
    '조에': '조에',
    '클리오': '클리오',
    '아르카나': '아르카나',
    '캡처': '캡처',

    # ═══ 쉐보레 (Chevrolet) ═══

    # 말리부
    '말리부': '말리부',
    '올 뉴 말리부': '말리부',
    '더 뉴 말리부': '말리부',

    # 스파크
    '스파크': '스파크',
    '더 뉴 스파크': '스파크',
    '더 넥스트 스파크': '스파크',

    # 올란도
    '올란도': '올란도',

    # 트레일블레이저
    '트레일블레이저': '트레일블레이저',
    '더 뉴 트레일블레이저': '트레일블레이저',

    # 트랙스
    '트랙스': '트랙스',
    '더 뉴 트랙스': '트랙스',
    '트랙스 크로스오버': '트랙스',

    # 크루즈
    '크루즈': '크루즈',
    '올 뉴 크루즈': '크루즈',
    '어메이징 뉴 크루즈': '크루즈',
    '크루즈5': '크루즈',
    '어메이징 뉴 크루즈5': '크루즈',

    # 이쿼녹스
    '이쿼녹스': '이쿼녹스',
    '더 넥스트 이쿼녹스': '이쿼녹스',

    # 콜로라도
    '콜로라도': '콜로라도',
    '리얼 뉴 콜로라도': '콜로라도',

    # 캡티바
    '캡티바': '캡티바',

    # 볼트
    '볼트 EV': '볼트',
    '뉴 볼트 EV': '볼트',
    '볼트 EUV': '볼트',

    # 아베오
    '아베오 세단': '아베오',
    '아베오 해치백': '아베오',
    '더 뉴 아베오 세단': '아베오',
    '더 뉴 아베오 해치백': '아베오',

    # 마티즈
    '마티즈 크리에이티브': '마티즈',
    'All New 마티즈': '마티즈',

    # 윈스톰
    '윈스톰': '윈스톰',
    '윈스톰 맥스': '윈스톰',

    # 단종/기타
    '알페온': '알페온',
    '임팔라': '임팔라',
    '트래버스': '트래버스',
    '타호': '타호',
    '라세티 프리미어': '라세티 프리미어',
    '올 뉴 카마로': '카마로',
    '더 뉴 카마로': '카마로',

    # ═══ 기타 단종 모델 ═══
    '오피러스 프리미엄': '오피러스',
    '뉴오피러스': '오피러스',
    '로체 이노베이션': '로체',
    '갤로퍼 2': '갤로퍼',
    '뉴 다마스': '다마스',
    '뉴 라보': '라보',
    '티코': '티코',
    '에스페로': '에스페로',
    '아카디아': '아카디아',
    '프레스토': '프레스토',
    '마르샤': '마르샤',
    '다이너스티': '다이너스티',
    '테라칸': '테라칸',
    '뉴 EF 쏘나타': '쏘나타',
    '포니투': '포니',
    '엑셀': '엑셀',
    '프라이드': '프라이드',
    '프라이드(신형)': '프라이드',
    '올 뉴 프라이드': '프라이드',
    '더 뉴 프라이드': '프라이드',
    '스테이츠맨': '스테이츠맨',
    'ST1': 'ST1',
    'PV5': 'PV5',
}


def main():
    df = pd.read_csv(CSV_PATH, encoding='utf-8')

    # 그룹 컬럼 생성 (매핑에 없으면 원래 이름 유지)
    df['모델_그룹'] = df['모델'].map(MODEL_GROUP_MAP).fillna(df['모델'])

    # 매핑 안 된 모델 확인
    unmapped = df[df['모델'] == df['모델_그룹']]['모델'].unique()
    unmapped_in_map = set(MODEL_GROUP_MAP.keys())
    truly_unmapped = [m for m in unmapped if m not in unmapped_in_map]

    # 결과: 그룹별 frequency
    result = (
        df.groupby('모델_그룹')
        .size()
        .reset_index(name='frequency')
        .sort_values('frequency', ascending=False)
        .reset_index(drop=True)
    )
    result.index += 1  # 1부터 시작

    print("=" * 50)
    print(f"  모델 그룹핑 결과  |  그룹 수: {len(result)}개")
    print("=" * 50)
    print(result.to_string())

    if truly_unmapped:
        print(f"\n※ 매핑 없이 원본명 유지된 모델 ({len(truly_unmapped)}개):")
        for m in sorted(truly_unmapped):
            count = df[df['모델'] == m].shape[0]
            print(f"  - {m} ({count}건)")

    return result


if __name__ == '__main__':
    result = main()
```
